```python
import pickle
import os
```

```python
# Define the folder where the data is saved
processed_data_folder = "../data/processed_data"
# Load the main data dictionary
with open(os.path.join(processed_data_folder, 'data_dic_for_yaren.pkl'), 'rb') as f:
    data_dic = pickle.load(f)
print(f"Loaded data_dic from {processed_data_folder}")
```

```python
"""
The dataset is organized as follows:
- data_dic: A dictionary where keys are 'mouse_recday'
- Each 'mouse_recday' key maps to another dictionary where keys are session identifiers (e.g., '0', '1', etc.)
- each session contains the data structures for that session. e.g. the location of the animal, or the neural firing rate arrays
- each session within a mouse_recday has the same number of neurons, as we tracked neurons across sessions
"""

print("keys in first session of first mouse_recday:", list(data_dic[list(data_dic.keys())[0]][list(data_dic[list(data_dic.keys())[0]].keys())[0]].keys()))
```
